## 1. 게이트 모델 알아보기

**게이트 모델(gated model)** 은 모델 주인이 사용 조건을 걸어 둔 저장소입니다. 라이선스에 동의하고 연락처를 보낸 계정만 파일을 받을 수 있습니다.

- Meta의 Llama 계열, 일부 음성·의료 모델이 게이트 모델입니다.
- 승인 전에는 모델 카드는 읽을 수 있어도 `Files and versions`의 가중치 파일은 받을 수 없습니다.
- 필요한 것은 셋입니다. Hugging Face 계정, 모델 주인의 승인, 코드에서 쓸 접근 토큰입니다.

로그인하지 않고 모델 페이지를 열면 아래 화면이 보입니다.

**주소:** `https://huggingface.co/meta-llama/Llama-Prompt-Guard-2-86M`

<img src="images/hf/hf_01_gated_logged_out.png" width="760" alt="로그인 전 Llama-Prompt-Guard-2-86M 페이지. 위쪽 모델 이름과 아래쪽 Log in, Sign Up 버튼에 빨간 상자">

**주요 요소**

- **모델 이름:** `meta-llama/Llama-Prompt-Guard-2-86M`처럼 `소유자/모델 이름` 모양입니다. 코드에서 모델을 부를 때 이 문자열을 그대로 씁니다.
- **License:** 이 모델은 `llama4` 라이선스를 따릅니다. 아래 상자에 라이선스 전문이 있습니다.
- **Log in / Sign Up:** 로그인해야 신청 양식이 열립니다. 계정이 없으면 Sign Up으로 먼저 가입합니다.

## 2. 사용 조건 열기

로그인한 뒤 아직 신청하지 않은 모델 페이지를 열면 사용 조건 상자가 보입니다. 아래 예시는 `Llama-Guard-3-8B`입니다.

**주소:** `https://huggingface.co/meta-llama/Llama-Guard-3-8B`

<img src="images/hf/hf_02_gate_box.png" width="760" alt="로그인 후 아직 신청하지 않은 Llama-Guard-3-8B 페이지의 사용 조건 상자. 맨 아래 Expand to review and access 버튼에 빨간 상자">

**주요 요소**

- **You need to agree to share your contact information:** 연락처를 모델 주인(Meta)에게 보내는 데 동의해야 받을 수 있다는 안내입니다.
- **LLAMA 3.1 COMMUNITY LICENSE AGREEMENT:** 이 모델의 라이선스입니다. `Expand to review`로 전문을 펼쳐 읽습니다.
- **Expand to review and access:** 누르면 라이선스 전문과 신청 양식이 함께 펼쳐집니다.

**Meta 모델은 묶음 단위로 승인됩니다.** 같은 라이선스를 쓰는 모델끼리 한 묶음(Gating Group Collection)입니다.

- Prompt Guard 2 86M·22M과 Llama Guard 4는 Llama 4 묶음입니다.
- Llama Guard 3 1B는 Llama 3.2 묶음, Llama Guard 3 8B는 Llama 3.1 묶음입니다.
- 한 모델이 승인되면 같은 묶음의 다른 모델도 따로 신청하지 않고 바로 받을 수 있습니다. 묶음이 다른 모델은 다시 신청합니다.

## 3. 신청 양식 채우기

`Expand to review and access`를 누르면 라이선스 아래에 양식이 나옵니다.

<img src="images/hf/hf_03_access_form.png" width="760" alt="Meta 모델 신청 양식. First Name, Last Name, Date of birth, Country, Affiliation, Job title 칸. Affiliation 입력 칸과, 맨 아래 동의 체크 상자와 Submit 버튼에 빨간 상자">

**칸마다 적을 것**

| 칸 | 뜻 | 적는 값 |
|---|---|---|
| First Name / Last Name | 이름 / 성 | 영문 실명(여권 표기처럼) |
| Date of birth | 생년월일 | 월/일/연도(`mm/dd/yyyy`) 순서입니다. 달력에서 고르면 편합니다 |
| Country | 나라 | South Korea |
| Affiliation | 소속 | 학교나 회사의 정식 영문 이름. 수강생이면 다니는 학교나 교육기관 이름 |
| Job title | 하는 일 | 목록에서 고릅니다. 학생이면 Student |

- **동의 체크 상자:** 라이선스에 동의하고 입력한 정보를 Meta가 수집하는 데 동의한다는 표시입니다. 체크한 뒤 Submit을 누릅니다.
- **Submit:** 신청을 보냅니다.
- 접속한 인터넷 주소로 추정한 나라·지역도 모델 주인에게 함께 전달됩니다(양식 아래 회색 안내).

**보내기 전에 확인할 것**

- Meta는 실명, 정확한 생년월일, 줄임말 없는 소속 이름을 요구하고 특수문자를 쓰지 말라고 안내합니다. 로그인 전 화면 맨 위 안내문에 이 조건이 있습니다.
- 보낸 뒤에는 양식을 고칠 수 없습니다. 안내를 따르지 않으면 이 모델과 다른 모델을 받지 못할 수 있다고 적혀 있습니다.

## 4. 승인 확인

승인되면 모델 페이지의 안내가 바뀝니다. 아래는 승인을 받은 `Llama-Prompt-Guard-2-86M` 페이지입니다.

<img src="images/hf/hf_04_access_granted.png" width="760" alt="승인 후 Llama-Prompt-Guard-2-86M 페이지. Gated model You have been granted access to this model 안내에 빨간 상자">

- **Gated model / You have been granted access to this model:** 이 계정이 파일을 받을 수 있다는 표시입니다. 이제 `Files and versions`에서 파일을 받거나 코드로 내려받을 수 있습니다.

신청한 모든 모델의 상태는 설정 화면 한 곳에서 봅니다. 오른쪽 위 프로필 사진을 누르고 Settings, 왼쪽 메뉴 Gated Repositories 순서로 엽니다.

**주소:** `https://huggingface.co/settings/gated-repos`

<img src="images/hf/hf_05_gated_repos.png" width="930" alt="Gated Repositories 화면. Meta's Llama 3.2 language models & evals 와 Llama 4 두 행, 종류 Gating Group Collection, 상태 ACCEPTED 에 빨간 상자">

**주요 요소**

- **Repo Name:** 신청한 저장소나 묶음 이름입니다. Meta 모델은 모델 하나가 아니라 묶음 이름으로 나옵니다.
- **Type:** 모델 하나면 `model`, 묶음이면 `Gating Group Collection`입니다.
- **Request Status:** 심사 중이면 `PENDING`, 승인되면 `ACCEPTED`, 거절되면 `REJECTED`입니다. `PENDING`인 동안에는 파일을 받을 수 없습니다.

## 5. 읽기 토큰 만들기

코드에서 게이트 모델을 받으려면 내 계정임을 알리는 **접근 토큰(access token)** 이 필요합니다. 브라우저 로그인과 별개입니다.

Settings 왼쪽 메뉴 Access Tokens에서 `Create new token`을 누릅니다.

**주소:** `https://huggingface.co/settings/tokens/new?tokenType=read`

<img src="images/hf/hf_06_token_new.png" width="930" alt="Create new Access Token 화면. Token type 의 Read 와 Create token 버튼에 빨간 상자">

**주요 요소**

- **Token type:** 모델을 내려받기만 하므로 `Read`를 고릅니다. 만든 뒤에는 바꿀 수 없습니다.
- `Write`는 내 저장소에 파일을 올릴 때, `Fine-grained`는 저장소마다 권한을 따로 줄 때 씁니다.
- **Token name:** 용도를 알아볼 이름을 적습니다(예: `day63-download`).
- **Create token:** 누르면 `hf_`로 시작하는 토큰 값이 나옵니다. 이 화면을 닫으면 다시 볼 수 없으니 바로 복사합니다.

**토큰 보관**

- 단원 폴더 `.env`에 `HF_TOKEN=복사한_토큰` 한 줄을 적습니다. `huggingface_hub`는 `HF_TOKEN` 환경변수를 자동으로 읽습니다.
- 토큰을 노트북 셀, 채팅, 깃허브에 붙여 넣지 않습니다. 노출되면 Access Tokens 목록에서 그 토큰을 삭제하고 새로 만듭니다.

## 6. 노트북에서 모델 내려받기

예시로 Prompt Guard 2의 작은 판인 `Llama-Prompt-Guard-2-22M`을 받습니다. 86M과 같은 Llama 4 묶음이라, 86M을 승인받은 계정은 따로 신청하지 않아도 됩니다.

먼저 `.env`의 토큰이 읽히는지, 어느 계정으로 로그인되는지 확인합니다.

```python
from pathlib import Path

from dotenv import find_dotenv, load_dotenv
from huggingface_hub import whoami

material_dir = Path(".")  # models 폴더가 있는 단원 폴더
model_dir = material_dir / "models"

# .env 의 HF_TOKEN 을 환경변수로 올립니다(huggingface_hub 가 이 값을 자동으로 씁니다)
load_dotenv(find_dotenv(usecwd=True))

# 토큰이 가리키는 계정 이름과 토큰 종류를 확인합니다
account = whoami()
print("계정:", account["name"])
print("토큰 종류:", account["auth"]["accessToken"]["role"])
```

- `LocalTokenNotFoundError`가 나면 토큰을 찾지 못한 것입니다. `.env`에 `HF_TOKEN` 줄이 있는지 확인합니다.
- `Invalid user token`이 나오면 토큰 값이 틀렸거나 지워진 토큰입니다.

다음으로 설정 파일 하나만 받아 승인 여부를 확인합니다. 가중치 전체를 받기 전에 작은 파일로 먼저 확인하면 실패를 빨리 알 수 있습니다.

```python
from huggingface_hub import hf_hub_download

MODEL_ID = "meta-llama/Llama-Prompt-Guard-2-22M"

# config.json 은 몇 KB 라서 승인 여부를 빠르게 확인할 수 있습니다
config_path = hf_hub_download(MODEL_ID, "config.json")
print("받은 파일:", config_path)
```

승인 전이면 `GatedRepoError`가 나고 메시지에 `Cannot access gated repo`가 보입니다. 이때는 Gated Repositories에서 상태가 `ACCEPTED`인지 확인합니다.

승인이 확인되면 모델 전체를 단원 폴더 `models/` 아래에 받습니다.

```python
from huggingface_hub import snapshot_download

# 저장소의 파일 전체를 local_dir 폴더에 받습니다
local_path = snapshot_download(MODEL_ID, local_dir=model_dir / "Llama-Prompt-Guard-2-22M")

# 받은 파일과 크기(MB)를 확인합니다
for path in sorted(Path(local_path).iterdir()):
    if path.is_file():
        print(path.name, round(path.stat().st_size / 1024 / 1024, 1), "MB")
```

- 받은 폴더는 교안 02의 `models/Llama-Prompt-Guard-2-86M`과 같은 모양입니다. 교안 02의 모델 경로만 바꾸면 같은 코드로 읽을 수 있습니다.
- `local_dir` 안에 생기는 `.cache/` 폴더는 다시 받을 때 쓰는 기록입니다. 깃허브에 올리지 않습니다.
- 라이선스 파일(`LICENSE` 또는 `LICENSE.txt`, `USE_POLICY.md`)도 함께 받아집니다. 모델을 다른 사람에게 나눠 줄 때는 이 파일을 같이 둡니다.
- 저장소에 `original/` 폴더(Meta 자체 형식의 사본)가 있으면 `ignore_patterns=["original/*"]`를 넣어 건너뜁니다. transformers로 읽을 때는 필요 없는 폴더입니다.
- 교안 02의 Llama Guard 3 1B가 이렇게 `original/`을 빼고 받은 모델입니다.
