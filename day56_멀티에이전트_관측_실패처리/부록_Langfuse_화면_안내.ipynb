# 📗 부록: Langfuse 화면 안내

Langfuse 웹 화면에서 자주 여는 페이지를 차례로 둘러봅니다. 교안 01의 5-4(Langfuse 화면 읽기)와 함께 봅니다. 페이지마다 무엇을 보는 곳인지, 화면의 주요 요소, 주소를 정리했습니다. 그림의 빨간 번호와 설명의 번호가 같은 곳을 가리킵니다. 화면 예시는 보험 청구 심사 팀(`claim-team`)과 시장 브리핑 팀(`briefing-team`)의 실행 기록입니다.

| 순서 | 페이지 | 여는 곳 |
|---|---|---|
| 1 | 들어가며: 주소 읽는 법 | - |
| 2 | 프로젝트 첫 화면 | Home |
| 3 | 실행 목록 | Tracing |
| 4 | trace 상세 | Tracing 목록에서 행 열기 |
| 5 | 모델 호출 상세 | trace 트리에서 ChatOpenAI 누르기 |
| 6 | observation 단위로 보기 | Tracing |
| 7 | 세션 목록과 세션 상세 | Sessions |
| 8 | 그 밖의 메뉴 | Prompts, Users, Dashboards, Scores 등 |

## 1. 들어가며

trace, observation, session과 observation 종류(generation 등)는 교안 01의 5-1에서, 태그는 5-2에서 정리했습니다. 이 부록은 그 기록을 화면에서 찾는 법만 다룹니다. 화면 예시의 태그 `v1`은 수정 전 실행, `v2-full-report`는 수정 후 실행입니다.

**주소 읽는 법**

모든 페이지 주소는 `https://cloud.langfuse.com/project/<프로젝트 ID>/...` 모양입니다.

- `https://cloud.langfuse.com` 부분은 `.env`의 `LANGFUSE_BASE_URL` 값입니다. 다른 지역이나 직접 설치한 서버를 쓰면 이 부분이 달라집니다.
- `<프로젝트 ID>`는 Langfuse에 로그인해 프로젝트를 연 뒤 주소창에서 `/project/` 바로 뒤에 오는 값입니다. 이 부록의 주소는 모두 이 자리를 `<프로젝트 ID>`로 적었습니다.
- `<trace id>`, `<세션 id>`, `<항목 id>`도 실제 값으로 바꿔 읽습니다. `<항목 id>`는 observation 하나의 id입니다.

## 2. 프로젝트 첫 화면(Home)

**주소:** `https://cloud.langfuse.com/project/<프로젝트 ID>`

프로젝트를 열면 처음 보이는 요약 대시보드입니다. 선택한 기간 동안 쌓인 기록의 양과 비용을 한눈에 봅니다.

<img src="images/langfuse/appendix/apx_home.png" width="1050" alt="Home 화면. 1 오른쪽 위 기간 선택(Past 1 day), 2 Traces 카드(이름별 trace 수: briefing-team 132, claim-team 39 등), 3 Model costs 카드(모델별 토큰과 비용), 4 Observations by time 그래프(DEBUG, ERROR, DEFAULT 선), 5 Model Usage 그래프에 빨간 상자">

**주요 요소**

1. **기간 선택:** 화면의 모든 숫자가 이 기간 기준입니다. 기본은 지난 하루(Past 1 day)이며, 오래전 실행이 안 보이면 기간을 늘립니다.
2. **Traces:** 기간 안의 trace 수와 trace 이름별 개수입니다. `briefing-team`, `claim-team`이 몇 번 실행됐는지 바로 보입니다.
3. **Model costs:** 모델별 토큰 수와 비용(USD)입니다.
4. **Observations by time:** 시간대별 observation 수를 수준(DEBUG, ERROR, DEFAULT)별 선으로 보여 줍니다. ERROR 선이 오른 시간대에는 예외로 끝난 단계가 많았습니다.
5. **Model Usage:** 시간대별 비용과 토큰 사용량입니다. 위쪽 탭으로 모델별·종류별 보기를 바꿉니다.

**쓰는 때:** 실습을 마친 뒤 오늘의 실행 횟수와 비용 확인.

## 3. 실행 목록(Tracing)

**주소:** `https://cloud.langfuse.com/project/<프로젝트 ID>/traces`

기록을 표로 모아 보는 곳입니다. 왼쪽 메뉴 이름은 **Tracing**입니다. 이 표는 observation을 한 행씩 보여 줍니다. 처음 열면 검색바에 `isRootObservation:true` 조건이 기본으로 들어 있어 각 trace의 맨 위 기록만 보입니다(**실행 한 번 = 한 행**). 이 조건을 지우면 trace 안의 단계가 모두 행으로 나옵니다(6절). 한 번 지우면 그 선택이 유지되므로, 실행 단위 목록으로 돌아가려면 조건을 직접 다시 넣습니다.

<img src="images/langfuse/appendix/apx_tracing_list.png" width="1050" alt="Tracing 목록. 1 검색바의 조건 칩 isRootObservation:true 와 name:=claim-team, 2 기간 선택, 3 Columns 버튼, 4 표 머리글(Start Time, Type, Name, Trace Name, Status, Latency, Trace Tags, Session ID), 5 실행 한 건의 행에 빨간 상자">

**주요 요소**

1. **검색바:** `필드:값` 조건을 입력하고 Enter를 누르면 조건 칩이 되어 목록을 거릅니다. 칩을 여러 개 넣으면 모두 만족하는 행만 남습니다.
2. **기간 선택:** 목록도 이 기간 안의 기록만 보여 줍니다.
3. **Columns:** 표에 보일 열을 고릅니다. Input, Output, Cost 열을 켜면 표에서 바로 입력·출력과 비용을 비교합니다.
4. **표 머리글:** 시작 시각(Start Time), 종류(Type), 이름(Name), 상태(Status), 걸린 시간(Latency) 같은 열이 보입니다. 태그(Trace Tags)와 세션 id(Session ID) 열은 Columns에서 켭니다(교안 01의 5-4). Status는 기록의 수준으로, 보통은 `Default`이고 예외로 끝난 단계는 `Error`입니다(4-4절).
5. **행:** 실행 한 건입니다. 누르면 오른쪽에 미리보기가 열리고, 패널 위 화살표로 이전·다음 실행을 넘겨 봅니다. 패널 맨 왼쪽 버튼은 trace 상세(4절)를 새 탭으로 엽니다.

**검색바에 자주 쓰는 조건**

| 입력 | 뜻 |
|---|---|
| `isRootObservation:true` | 각 trace의 맨 위 기록만 봅니다(실행 단위 목록, 처음 열 때 기본값) |
| `name:=claim-team` | 이름이 정확히 `claim-team`인 기록만 봅니다 |
| `traceTags:v1` | 태그 `v1`이 붙은 trace의 기록만 봅니다 |
| `sessionId:claim-review-2026-10-08-CLM-2101` | 한 세션의 기록만 봅니다 |
| `type:GENERATION` | 모델 호출 기록만 봅니다(6절) |

조건을 넣으면 주소도 `.../traces?filter=...` 꼴로 바뀝니다. 이 주소를 저장하거나 공유하면 같은 조건의 목록을 다시 엽니다.

**쓰는 때:** 수정 전후 실행을 `traceTags:v1`과 `traceTags:v2-full-report`로 나눠 걸러 걸린 시간과 결과 비교(교안 01의 6절).

## 4. trace 상세

**주소:** `https://cloud.langfuse.com/project/<프로젝트 ID>/traces/<trace id>`

실행 한 번을 펼쳐 보는 페이지입니다. 왼쪽에 단계 목록, 오른쪽에 고른 단계의 입력과 출력이 보입니다. 보기 방식은 Tree, Timeline, Graph 세 가지입니다.

| 보기 | 무엇을 보나 |
|---|---|
| Tree | 단계가 실행 순서대로 포개진 나무 모양 |
| Timeline | 단계마다 언제 시작해 얼마나 걸렸는지 |
| Graph | 노드 사이 연결과 노드별 실행 횟수 |

### 4-1. Tree 보기와 오른쪽 패널

<img src="images/langfuse/appendix/apx_trace_tree.png" width="1050" alt="trace 상세의 Tree 보기. 1 머리 요약 줄(1m 36s, 비용, 토큰, session, 태그 day56과 v1), 2 Tree·Timeline·Graph 탭, 3 Formatted·Raw 전환, 4 고른 단계의 Input과 Output에 빨간 상자">

**주요 요소**

1. **머리 요약 줄:** 전체 걸린 시간, 비용, 토큰 수, 태그입니다. **session ↗**을 누르면 이 trace가 속한 세션 상세(7절)로 갑니다.
2. **보기 탭:** Tree, Timeline, Graph를 바꿉니다.
3. **Formatted / Raw:** Formatted는 표로 정리해 보여 주고, Raw는 저장된 JSON 그대로 보여 줍니다. 값이 잘리거나 중첩이 깊으면 Raw로 봅니다.
4. **Preview(Input / Output):** 왼쪽에서 고른 단계가 받은 입력과 내놓은 출력입니다. 맨 위 `claim-team`을 고르면 팀 전체의 처음 State와 마지막 State가 보입니다.

**트리 읽는 법:** 워커 노드(`claim_worker`) 아래에 에이전트(`claim_agent`)가, 그 아래에 모델 호출(`model` > ChatOpenAI)과 도구 호출(`tools` > `get_claim`)이 포개집니다. `supervisor` 아래의 RunnableSequence, ChatPromptTemplate, RunnableLambda는 슈퍼바이저 판단 체인의 단계(프롬프트, 모델 호출, 결과 변환)가 자동으로 기록된 것입니다. 오른쪽 위 Add to, Annotate, Comment는 기록을 데이터셋에 넣거나 점수·메모를 다는 기능이며, day56에서는 쓰지 않습니다.

**쓰는 때:** 최종 답변이 틀렸을 때, 어느 단계에서 정보가 빠졌는지 단계별 입력과 출력을 차례로 열어 확인.

### 4-2. Timeline 보기

<img src="images/langfuse/appendix/apx_trace_timeline.png" width="1050" alt="Timeline 보기. 1 확대·축소·화면 맞춤 버튼과 시간 눈금(0ms부터 1m 15s), 2 claim_worker 한 번의 실행 막대 묶음(claim_agent, model, tools, get_claim)에 빨간 상자">

**주요 요소**

1. **시간 눈금과 확대 버튼:** 위쪽 눈금이 실행 시작부터 지난 시간입니다. `-`, `+`로 확대·축소하고, 네모 버튼으로 전체 실행이 한 화면에 들어오게 맞춥니다. 막대 옆 글자는 그 단계의 걸린 시간과 비용입니다.
2. **막대:** 단계마다 시작 위치와 길이로 언제, 얼마나 걸렸는지 보입니다. 워커 막대 안에서 모델 호출 막대가 대부분을 차지하면 느린 원인은 모델 호출입니다.

**쓰는 때:** 실행이 느릴 때 시간을 잡아먹는 단계 찾기, 병렬로 보낸 작업이 실제로 겹쳐 돌았는지 확인.

### 4-3. Graph 보기

<img src="images/langfuse/appendix/apx_trace_graph.png" width="1050" alt="Graph 보기. 1 Aggregated·Expanded 전환, 2 claim_worker (7/7) 노드, 3 확대·축소·화면 맞춤 버튼에 빨간 상자">

**주요 요소**

1. **Aggregated / Expanded:** Aggregated는 같은 이름의 노드를 하나로 합쳐 보여 주고, Expanded는 실행된 차례대로 모두 펼칩니다.
2. **노드와 숫자:** `claim_worker (7/7)`처럼 괄호 `(왼쪽/오른쪽)`의 **오른쪽 숫자가 이 trace에서 그 노드가 실행된 횟수**입니다. 왼쪽 숫자는 그 노드를 누를 때 오른쪽 패널에 열리는 실행 기록을 가리키는 표시입니다. 처음에는 오른쪽 숫자와 같고, 같은 노드를 다시 누를 때마다 하나씩 줄면서 다른 실행 기록으로 넘어갑니다(5/5, 4/5, … 1/5 다음은 다시 5/5). 그래서 처음 화면에서는 두 숫자가 같습니다. 한 번만 실행된 노드에는 괄호가 붙지 않습니다. 화살표는 노드 사이의 연결입니다. `tools`, `model`은 팀 노드가 아니라 워커 안 에이전트의 단계입니다.
3. **확대 버튼:** 그래프를 키우거나 줄이고, 네모 버튼으로 화면에 맞춥니다.

**쓰는 때:** 슈퍼바이저가 한 워커만 여러 번 부르는지 한눈에 확인. 그림은 CLM-2101의 v1 실행입니다. `claim_worker`가 7번 불렸고, 숫자 없이 보이는 `terms_worker`, `reply_writer`는 한 번씩 불렸습니다. 청구 조회는 한 번이면 끝나므로 이 실행은 같은 일을 반복했습니다.

### 4-4. 실패한 단계 보기

<img src="images/langfuse/appendix/apx_trace_error.png" width="1050" alt="실패가 있는 briefing-team trace(태그 lesson02, v2-retry). 1 트리에서 Error 표시가 붙은 collect_prices 두 번과 성공한 세 번째, 2 오른쪽 패널의 Error 표시, 3 Error 상자 안의 오류 메시지 '가격 API 연결 끊김'에 빨간 상자. 그 아래 Output 칸은 undefined">

**주요 요소**

1. **Error 표시:** 예외로 끝난 단계는 트리에서 이름 옆에 빨간 **Error**가 붙습니다. 그림은 교안 02의 4-2 재시도 실행으로, `collect_prices`가 두 번 실패하고 세 번째에 성공한 기록입니다.
2. **오른쪽 패널의 Error:** 고른 단계의 시각, 걸린 시간 옆에도 Error가 표시됩니다.
3. **Error 상자:** 예외 메시지가 Input 위에 따로 보입니다. 실패한 단계는 내놓은 값이 없어 Output이 `undefined`입니다.

**쓰는 때:** 실행이 중간에 멈췄을 때 멈춘 단계와 오류 내용, 재시도 횟수 확인.

## 5. 모델 호출 상세

**주소:** `https://cloud.langfuse.com/project/<프로젝트 ID>/traces/<trace id>?observation=<항목 id>`

trace 트리에서 **ChatOpenAI**처럼 모델 호출 기록(generation)을 누르면 오른쪽 패널이 그 호출의 상세로 바뀌고, 주소 끝에 `?observation=<항목 id>`가 붙습니다. 이 주소를 열면 같은 단계가 고른 상태로 바로 열립니다.

<img src="images/langfuse/appendix/apx_generation.png" width="1050" alt="모델 호출 상세. 1 트리에서 고른 ChatOpenAI, 2 시각·걸린 시간·비용·토큰 수·모델 이름(gpt-5.6-luna) 줄, 3 Preview·Attributes·Scores·Log View 탭, 4 Tools(모델에 준 도구 get_claim과 called 표시), 5 System과 User 메시지, 6 Assistant 응답(도구 호출 get_claim, claim_id CLM-2101)에 빨간 상자">

**주요 요소**

1. **고른 모델 호출:** `claim_worker` 안의 첫 `model` 단계 아래 ChatOpenAI입니다.
2. **요약 줄:** 호출 시각, 걸린 시간, 비용, 토큰 수, 모델 이름입니다.
3. **탭:** Preview는 메시지, Attributes는 모델 설정 같은 속성, Scores는 이 호출에 붙은 점수를 보여 줍니다. Log View는 trace 안 모든 단계의 기록을 위에서 아래로 이어 붙여 보여 주므로, 긴 trace에서 특정 값을 Ctrl+F(맥은 Cmd+F)로 찾을 때 씁니다.
4. **Tools:** 모델에 건넨 도구 목록입니다. **called**는 이 호출에서 모델이 그 도구를 부르기로 했다는 뜻입니다.
5. **System / User:** 모델이 실제로 받은 메시지입니다. 프롬프트 템플릿에 값이 채워진 최종 문장이 보입니다.
6. **Assistant:** 모델의 응답입니다. 그림에서는 답 대신 `get_claim(claim_id="CLM-2101")` 도구 호출을 냈습니다. 추론 모델은 생각 과정이 `Thinking (redacted)`로 가려져 보입니다.

**쓰는 때:** 워커가 엉뚱한 일을 했을 때, 프롬프트에 필요한 정보가 실제로 들어갔는지 System과 User 칸에서 확인.

## 6. observation 단위로 보기

**주소:** `https://cloud.langfuse.com/project/<프로젝트 ID>/traces` (검색바에 `type:GENERATION` 등)

Langfuse v4에는 observation 목록 메뉴가 따로 없습니다. 같은 Tracing 목록에서 `isRootObservation:true`를 지우고 종류(Type)로 거르면 trace를 열지 않고도 단계 기록만 모아 봅니다.

<img src="images/langfuse/appendix/apx_observations.png" width="1050" alt="Tracing 목록을 observation 단위로 거른 화면. 1 검색바 조건 type:GENERATION 과 sessionId:claim-review-2026-10-08-CLM-2101, 2 왼쪽 Filters 패널의 Type 항목(CHAIN, GENERATION, RETRIEVER, TOOL, AGENT와 개수), 3 Type과 Name 열에 모델 호출 ChatOpenAI만 남은 행들, 4 목록 아래 Total ≈ 104에 빨간 상자">

**주요 요소**

1. **조건:** `type:GENERATION sessionId:claim-review-2026-10-08-CLM-2101`은 한 세션의 모델 호출만 남깁니다.
2. **Filters 패널:** 표 왼쪽 위 아이콘으로 엽니다. Type 항목은 종류별 개수를 보여 주며, 이름을 누르면 그 종류만, 체크박스를 누르면 켜고 끄기입니다.
3. **Type / Name 열:** 아이콘으로 종류를 구분합니다. 종류별 뜻은 교안 01의 5-1 표와 같습니다.
4. **Total:** 조건에 맞는 행 수입니다. 이 조건에서는 이 세션의 모델 호출 수이며, `≈`가 붙으면 어림값입니다.

**쓰는 때:** 한 작업에서 모델을 몇 번 불렀는지, 그중 오래 걸린 호출이 무엇인지 Latency 열로 정렬해 확인.

## 7. 세션 목록과 세션 상세

**주소:** `https://cloud.langfuse.com/project/<프로젝트 ID>/sessions` (목록), `.../sessions/<세션 id>` (상세)

세션은 같은 세션 id를 단 trace들의 묶음입니다. day56 예제는 청구 하나를 심사하는 작업 하나에 세션 하나(`claim-review-2026-10-08-CLM-2101`)를 쓰므로, 같은 청구를 다시 실행한 기록이 한 세션에 모입니다. 수정 후 실행(v2)은 새 작업이므로 이름 끝에 `-v2`를 붙인 다른 세션입니다(교안 01의 6절).

<img src="images/langfuse/appendix/apx_sessions.png" width="1050" alt="Sessions 목록. 1 검색바 조건 id:claim-review-2026-10-08-CLM, 2 ID 열의 세션 이름들(청구별 세션과 각각의 -v2 세션), 3 Duration 열, 4 Traces 수와 Total Cost 열에 빨간 상자">

**세션 목록의 주요 요소**

1. **검색바:** `claim-review-2026-10-08-CLM`처럼 글자를 넣으면 `id:claim-review-2026-10-08-CLM` 칩이 되어 그 글자가 들어간 세션만 남습니다.
2. **ID:** 세션 id입니다. 누르면 세션 상세가 열립니다.
3. **Duration:** 세션의 첫 trace 시작부터 마지막 trace 끝까지의 시간입니다. 같은 세션으로 나중에 다시 실행하면 그만큼 길어집니다.
4. **Traces / Total Cost:** 세션에 묶인 trace 수와 비용 합계입니다.

<img src="images/langfuse/appendix/apx_session_detail.png" width="1050" alt="세션 상세. 1 Session 이름 claim-review-2026-10-08-CLM-2101, 2 보기 선택(All observations with I/O)과 Filter observations, 3 Total traces 4와 Total cost, 4 trace 카드의 제목 링크 claim-team(trace id), 5 카드 맨 위 칸인 claim-team의 Input과 Output에 빨간 상자">

**세션 상세의 주요 요소**

1. **세션 이름:** 주소의 `<세션 id>`와 같습니다. 옆 아이콘으로 세션 주소를 공유하거나 id를 복사합니다.
2. **보기와 필터:** 기본 보기(All observations with I/O)는 입력과 출력이 모두 있는 기록만 보여 줍니다.
3. **합계:** 세션의 trace 수와 비용 합계입니다.
4. **trace 카드:** trace마다 카드 하나가 시간 순서대로 아래로 이어집니다. 제목을 누르면 그 trace 상세(4절)가 열립니다.
5. **Input / Output:** 카드 안에는 입력과 출력이 있는 단계마다 한 칸씩 이어집니다. 맨 위 칸이 팀 전체(`claim-team`)의 입력과 출력입니다. 스크롤을 내리며 실행마다 결과가 어떻게 달라졌는지 봅니다.

**쓰는 때:** 한 작업에 쌓인 실행 수 확인, 실패한 실행 뒤에 같은 세션으로 다시 실행한 기록이 이어졌는지 확인.

## 8. 그 밖의 메뉴

아래 메뉴는 무엇을 하는 곳인지만 짧게 정리합니다. 주소는 모두 `https://cloud.langfuse.com/project/<프로젝트 ID>/` 뒤에 붙는 부분입니다.

| 메뉴 | 주소 끝 | 무엇을 하는 곳인가 |
|---|---|---|
| Prompts | `prompts` | 프롬프트를 코드 밖에 버전으로 저장하고, 코드는 이름과 라벨(production 등)로 불러옵니다(22일차) |
| Users | `users` | trace에 사용자 id를 붙이면 사용자별 실행 수와 비용을 봅니다. 이 프로젝트는 사용자 id를 붙이지 않아 비어 있습니다 |
| Dashboards | `dashboards` | 기록을 그래프로 모은 대시보드입니다. 2절의 Home도 그중 하나이며, 비용·지연 시간·에이전트 도구 사용을 따로 모은 기본 대시보드도 있습니다 |
| Scores | `scores` | 실행 결과에 붙인 점수(사용자 평가, 모델 평가, 사람 검토)를 모아 봅니다 |
| Evaluators | `evals` | 모델이나 코드로 trace에 점수를 자동으로 매기는 규칙을 만듭니다 |
| Human Annotation | `annotation-queues` | 사람이 기록을 검토하고 점수를 매기도록 대기열을 만듭니다 |
| Datasets / Experiments | `datasets`, `experiments` | 질문과 기대 답 모음(데이터셋)을 만들고, 새 버전을 그 데이터셋으로 실험해 비교합니다 |
| Playground | `playground` | 프롬프트와 모델 설정을 화면에서 바로 바꿔 호출해 봅니다 |
| Alerts | `alerts` | 조건을 정해 두고 넘으면 알림을 받습니다 |
| Settings | `settings` | 프로젝트 설정입니다. API 키는 `settings/api-keys`에서 만듭니다 |

Settings의 API Keys 화면에는 비밀 키가 보일 수 있습니다. 화면을 공유하거나 캡처할 때는 이 화면을 피하고, 키는 `.env`에만 둡니다.
